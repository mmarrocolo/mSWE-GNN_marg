# Visualize fine-tuned mSWE-GNN — Ahr river 100 m
Loads the fine-tuned checkpoint and runs a full rollout on the 100 m additionalsrc dataset.

```python
import sys, os

_proj_db = r'C:\Users\marrocol\AppData\Local\miniforge3\envs\mswe-gnn\Lib\site-packages\pyproj\proj_dir\share\proj'
os.environ.setdefault('PROJ_DATA', _proj_db)
os.environ.setdefault('PROJ_LIB',  _proj_db)

try:
    REPO_ROOT = os.path.abspath(os.path.join(os.path.dirname(__vsc_ipynb_file__), '..'))
except NameError:
    REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))

os.chdir(REPO_ROOT)
if REPO_ROOT not in sys.path:
    sys.path.insert(0, REPO_ROOT)

import torch
import wandb
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from utils.load import read_config
from utils.miscellaneous import get_model, fix_dict_in_config
from utils.dataset import create_model_dataset, get_temporal_test_dataset_parameters, to_temporal_dataset
from utils.visualization import PlotRollout
from training.train import LightningTrainer

torch.backends.cudnn.deterministic = True
torch.set_float32_matmul_precision('high')
print('Repo root:', REPO_ROOT)
```

## Paths

```python
import glob

candidates = sorted(
    glob.glob(os.path.join(REPO_ROOT, 'lightning_logs', '**', '*.ckpt'), recursive=True) +
    glob.glob(os.path.join(REPO_ROOT, 'results', '*.h5'))
)

print(f"{'File':<55}  {'SizeMB':>7}")
for p in candidates:
    mb = os.path.getsize(p) / 1e6
    print(f"{os.path.relpath(p, REPO_ROOT):<55}  {mb:>7.1f}")

CHECKPOINT   = os.path.join(REPO_ROOT, 'lightning_logs', 'best_changed_lossandCSI.ckpt')
CONFIG       = 'config_finetune_100m_velocity.yaml'
DATASET_NAME = 'ahr_river_v03_marg_additionalsrc_velocity_100m_cutpolygon_warmstart'

print('\nActive checkpoint:', os.path.relpath(CHECKPOINT, REPO_ROOT))
print('Dataset:          ', DATASET_NAME)
```

## Load config and dataset

```python
cfg = read_config(CONFIG)
cfg['dataset_parameters']['test_dataset_name']  = DATASET_NAME
cfg['dataset_parameters']['train_dataset_name'] = DATASET_NAME

wandb.finish()  # ensure no stale wandb run reuses old config
wandb.init(mode='disabled', project='mswe-gnn', config=cfg)
fix_dict_in_config(wandb)
config = wandb.config

# confirm correct dataset is loaded
print('Loading dataset:', config.dataset_parameters['test_dataset_name'])

device = torch.device('cpu')

_, _, test_dataset, scalers = create_model_dataset(
    scalers=config.scalers, device=device,
    **config.dataset_parameters,
    **config.selected_node_features,
    **config.selected_edge_features
)

temporal_test_dataset_parameters = get_temporal_test_dataset_parameters(
    config, config.temporal_dataset_parameters
)

print('Test size:', len(test_dataset))
print('WD shape: ', test_dataset[0].WD.shape)
print('WD at t=0 (max):', test_dataset[0].WD[:, 0].max().item(), 'm  ← should be >0 for warmstart')
```

## Inputs before and after scaling

```python
import pickle

# Load raw (unscaled) dataset directly from PKL
pkl_path = os.path.join(REPO_ROOT, 'database', 'datasets', 'test',
                        f'{DATASET_NAME}.pkl')
with open(pkl_path, 'rb') as f:
    raw_data = pickle.load(f)
raw = raw_data[0]

dem_raw  = raw.DEM.numpy().flatten()
area_raw = raw.area.numpy().flatten()

# Recover raw edge length via inverse_transform (handle list or scalar scaler)
el_scaler = scalers.get('edge_length_scaler')
if isinstance(el_scaler, list):
    el_scaler = el_scaler[0]
edge_scaled = test_dataset[0].edge_attr[:, 0].numpy().reshape(-1, 1)
edge_raw = el_scaler.inverse_transform(edge_scaled).flatten() if el_scaler is not None else edge_scaled.flatten()

fig, axes = plt.subplots(3, 2, figsize=(12, 10))
fig.suptitle('Input feature distributions — raw vs standard scaled', fontsize=13)

for ax_raw, ax_sc, raw_vals, name, unit in zip(
        axes[:, 0], axes[:, 1],
        [dem_raw, area_raw, edge_raw],
        ['DEM', 'Cell area', 'Edge length'],
        ['m', 'm²', 'm']):
    scaled_vals = (raw_vals - raw_vals.mean()) / raw_vals.std()
    ax_raw.hist(raw_vals, bins=60, color='steelblue', alpha=0.7)
    ax_raw.set_title(f'{name} — raw [{unit}]  μ={raw_vals.mean():.1f}, σ={raw_vals.std():.1f}')
    ax_sc.hist(scaled_vals, bins=60, color='coral', alpha=0.7)
    ax_sc.set_title(f'{name} — standardized  μ≈0, σ≈1')
    ax_sc.axvline(0, color='k', lw=0.8, ls='--')

plt.tight_layout()
plt.show()
```

## Load fine-tuned model

```python
temporal_test_dataset = to_temporal_dataset(
    test_dataset, rollout_steps=-1, **temporal_test_dataset_parameters
)

num_node_features = temporal_test_dataset[0].x.size(-1)
num_edge_features = temporal_test_dataset[0].edge_attr.size(-1)

model_parameters = dict(config.models)
model_type = model_parameters.pop('model_type')
if model_type == 'MSGNN':
    model_parameters['num_scales'] = test_dataset[0].mesh.num_meshes

# Auto-detect architecture from checkpoint to handle mismatches
_ckpt = torch.load(CHECKPOINT, map_location='cpu', weights_only=False)
_sd   = _ckpt['state_dict']
_hid  = _sd['model.edge_encoder.0.weight'].shape[0]
_K    = sum(1 for k in _sd if 'gnn_processor.0.filter_matrix.' in k and k.endswith('.weight')) - 1
if _hid != model_parameters.get('hid_features') or _K != model_parameters.get('K'):
    print(f"Checkpoint architecture differs from config — overriding: hid_features={_hid}, K={_K}")
    model_parameters['hid_features'] = _hid
    model_parameters['K']            = _K

model = get_model(model_type)(
    num_node_features=num_node_features,
    num_edge_features=num_edge_features,
    previous_t=temporal_test_dataset_parameters['previous_t'],
    device=device,
    **model_parameters
).to(device)

plmodule_kwargs = {
    'model': model,
    'lr_info': config['lr_info'],
    'trainer_options': config.trainer_options,
    'temporal_test_dataset_parameters': temporal_test_dataset_parameters
}

plmodule = LightningTrainer.load_from_checkpoint(
    CHECKPOINT, map_location=device, **plmodule_kwargs
)
model = plmodule.model.to(device)
model.eval()
print('Model loaded from:', CHECKPOINT)
```

## Visualise rollout

```python
plot_rollout = PlotRollout(
    model,
    test_dataset[0],
    scalers=scalers,
    **temporal_test_dataset_parameters
)
print('Total time steps:', plot_rollout.real_rollout.shape[-1])
```

```python
# DEM, ground truth, prediction, difference at the last time step
fig = plot_rollout.explore_rollout(scale=0, time_step=-1)
plt.show()
```

```python
n_steps = plot_rollout.real_rollout.shape[-1]
plot_times = [0, 12, 48, 72, 96, n_steps - 1]
plot_rollout.compare_h_rollout(plot_times=plot_times, scale=3)
plt.show()
```

```python
# CSI over time
thresholds = [0.05, 0.30]

fig, ax = plt.subplots(figsize=(10, 4))
for thr in thresholds:
    csi = plot_rollout._get_CSI(water_threshold=thr).detach().cpu().numpy()
    if csi.ndim > 1:
        csi = np.nanmean(csi, axis=0)
    ax.plot(csi, label=f'CSI @ {thr:.2f} m')

ax.set_xlabel('Time step')
ax.set_ylabel('CSI')
ax.set_ylim(0, 1)
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

## Metrics

```python
rollout_loss = plot_rollout._get_rollout_loss(type_loss='RMSE')
loss_mean = rollout_loss.mean(0)
if loss_mean.ndim == 0:
    print(f'RMSE WD: {loss_mean.item():.4f} m')
    loss_mean = loss_mean.unsqueeze(0)
else:
    print(f'RMSE WD: {loss_mean[0].item():.4f} m')
if loss_mean.shape[0] > 1:
    print(f'RMSE V:  {loss_mean[1:].mean().item():.4f} m²/s')

csi_005 = plot_rollout._get_CSI(water_threshold=0.05).nanmean().item()
csi_03  = plot_rollout._get_CSI(water_threshold=0.30).nanmean().item()
print(f'CSI @ 0.05 m: {csi_005:.4f}')
print(f'CSI @ 0.30 m: {csi_03:.4f}')
```

```python
# --- Diagnostic: does the model work when flood has already started? ---
# time_start=0 → input WD at t=0,1,2 → all dry (simulation starts dry)
# time_start=N → input WD at t=N,N+1,N+2 → flood may already be present
from training.train import rollout_test
from utils.dataset import to_temporal

model.eval()
print(f"{'time_start':>12}  {'n_input_wet':>12}  {'max_pred_t5':>12}  {'max_pred_t20':>12}  {'CSI@0.05':>10}")
print('-' * 65)

for ts in [0, 3, 5, 10, 15, 20, 30]:
    params = dict(temporal_test_dataset_parameters)
    params['time_start'] = ts
    try:
        _td = to_temporal(test_dataset[0], rollout_steps=-1, **params)[0]
    except Exception:
        print(f"{ts:>12}  skipped (not enough steps)")
        continue

    # count wet nodes in initial condition
    n_wet_init = (_td.x[:, -2:].abs() > 0).any(dim=1).sum().item()  # non-zero WD or V in input

    with torch.no_grad():
        preds = rollout_test(model, _td)  # [N, vars, T]

    mp5  = preds[:, 0, min(5,  preds.shape[-1]-1)].max().item()
    mp20 = preds[:, 0, min(20, preds.shape[-1]-1)].max().item()

    real = _td.y
    T = min(preds.shape[-1], real.shape[-1])
    TP = ((preds[:, 0, :T] > 0.05) & (real[:, 0, :T] > 0.05)).sum().item()
    FP = ((preds[:, 0, :T] > 0.05) & (real[:, 0, :T] <= 0.05)).sum().item()
    FN = ((preds[:, 0, :T] <= 0.05) & (real[:, 0, :T] > 0.05)).sum().item()
    csi = TP / (TP + FP + FN) if (TP + FP + FN) > 0 else float('nan')

    print(f"{ts:>12}  {n_wet_init:>12}  {mp5:>12.4f}  {mp20:>12.4f}  {csi:>10.4f}")
```

## Comparison: new model (all-scale loss) vs old model (scale-0 loss)

```python
from utils.miscellaneous import get_rollout_loss, get_CSI

CHECKPOINT_OLD = os.path.join(REPO_ROOT, 'lightning_logs', 'finetuned_ahr_best_scale0loss.ckpt')

# load old model
_ckpt_old = torch.load(CHECKPOINT_OLD, map_location='cpu', weights_only=False)
_sd_old   = _ckpt_old['state_dict']
_hid_old  = _sd_old['model.edge_encoder.0.weight'].shape[0]
_K_old    = sum(1 for k in _sd_old if 'gnn_processor.0.filter_matrix.' in k and k.endswith('.weight')) - 1

mp_old = dict(config.models)
mp_old.pop('model_type')
if model_type == 'MSGNN':
    mp_old['num_scales'] = test_dataset[0].mesh.num_meshes
mp_old['hid_features'] = _hid_old
mp_old['K']            = _K_old

model_old = get_model(model_type)(
    num_node_features=num_node_features,
    num_edge_features=num_edge_features,
    previous_t=temporal_test_dataset_parameters['previous_t'],
    device=device,
    **mp_old
).to(device)

plmodule_old = LightningTrainer.load_from_checkpoint(
    CHECKPOINT_OLD, map_location=device,
    **{**plmodule_kwargs, 'model': model_old}
)
model_old = plmodule_old.model.to(device)
model_old.eval()
print('Old model loaded from:', os.path.relpath(CHECKPOINT_OLD, REPO_ROOT))
```

```python
plot_rollout_old = PlotRollout(
    model_old,
    test_dataset[0],
    scalers=scalers,
    **temporal_test_dataset_parameters
)

# extract SFINCS-scale (finest) rollouts for fair comparison
from utils.dataset import separate_multiscale_node_features
node_ptr = test_dataset[0].node_ptr

def get_finest_rollout(pr):
    pred = separate_multiscale_node_features(pr.predicted_rollout, node_ptr)[-1]
    real = separate_multiscale_node_features(pr.real_rollout,      node_ptr)[-1]
    return pred, real

pred_new, real = get_finest_rollout(plot_rollout)
pred_old, _    = get_finest_rollout(plot_rollout_old)

rmse_new  = get_rollout_loss(pred_new, real, type_loss='RMSE').mean().item()
rmse_old  = get_rollout_loss(pred_old, real, type_loss='RMSE').mean().item()
csi05_new = get_CSI(pred_new, real, water_threshold=0.05).nanmean().item()
csi05_old = get_CSI(pred_old, real, water_threshold=0.05).nanmean().item()
csi30_new = get_CSI(pred_new, real, water_threshold=0.30).nanmean().item()
csi30_old = get_CSI(pred_old, real, water_threshold=0.30).nanmean().item()

print(f"{'Metric':<20} {'New (all-scale loss)':>22} {'Old (scale-0 loss)':>20}")
print('-' * 64)
print(f"{'RMSE WD [m]':<20} {rmse_new:>22.4f} {rmse_old:>20.4f}")
print(f"{'CSI @ 0.05 m':<20} {csi05_new:>22.4f} {csi05_old:>20.4f}")
print(f"{'CSI @ 0.30 m':<20} {csi30_new:>22.4f} {csi30_old:>20.4f}")
```
